```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter
import locale
locale.setlocale(locale.LC_TIME, 'pt_BR.UTF-8')
```

```python
dict_faturamento = {
    'data_ref': [
        '2023-01-01', 
        '2020-02-01', 
        '2021-03-01', 
        '2022-04-01', 
        '2023-05-01',
        '2023-06-01', 
        '2020-07-01', 
        '2021-08-01', 
        '2022-09-01', 
        '2023-10-01',
        '2022-11-01', 
        '2023-12-01',
        ],
    'valor': [
        400000, 
        890000, 
        760000, 
        430000, 
        920000,
        340000, 
        800000, 
        500000, 
        200000, 
        900000,
        570000, 
        995000,
        ]
}

df_faturamento = pd.DataFrame.from_dict(dict_faturamento)
```

```python
print(df_faturamento.valor.mean())

df_faturamento['data_ref'] = pd.to_datetime(df_faturamento['data_ref'])
df_faturamento['label'] = df_faturamento['data_ref'].dt.strftime('%b/%Y')

def format_milhar(x, _):
  return f'{x/1_000_000:.1f}M'

plt.figure(figsize=(12, 6))
plt.bar(df_faturamento['label'], df_faturamento['valor'])
plt.xticks(rotation=45)
plt.ylabel('Valor (R$)', fontsize=12)
plt.title('Faturamento por Mês', fontsize=14)
plt.gca().yaxis.set_major_formatter(FuncFormatter(format_milhar))
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
plt.plot(df_faturamento['label'], df_faturamento['valor'], marker='o')
plt.xticks(rotation=45)
plt.ylabel('Valor (R$)', fontsize=12)
plt.title('Faturamento por Mês', fontsize=14)
plt.gca().yaxis.set_major_formatter(FuncFormatter(format_milhar))
plt.tight_layout()
plt.show()
```
